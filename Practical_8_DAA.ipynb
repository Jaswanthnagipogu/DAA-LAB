# IMPLEMENTATION OF GRAPH AND SEARCHING(DFS AND BFS)

```python
from queue import Queue

# Graph
graph = {
    '1': ['2', '3'],
    '2': ['1', '4', '5'],
    '3': ['1', '6'],
    '4': ['2'],
    '5': ['2'],
    '6': ['3'],
}

# DFS
def dfs(start):
    visited = []
    stack = [start]

    print("DFS Traversal:")

    while stack:
        node = stack.pop()

        if node not in visited:
            print(node, end=" ")
            visited.append(node)

            for i in reversed(graph[node]):
                if i not in visited:
                    stack.append(i)

# BFS
def bfs(start):
    visited = []
    q = Queue()

    visited.append(start)
    q.put(start)

    print("\nBFS Traversal:")

    while not q.empty():
        node = q.get()
        print(node, end=" ")

        for i in graph[node]:
            if i not in visited:
                visited.append(i)
                q.put(i)

# Main
start = input("Enter starting node: ")

dfs(start)
bfs(start)
```
