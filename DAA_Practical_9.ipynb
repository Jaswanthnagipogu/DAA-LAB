
# Prim's algorithm



```python
# Prim's Algorithm

graph = [
    [0, 3, 6, 0],
    [3, 0, 2, 4],
    [6, 2, 0, 0],
    [0, 4, 0, 0]
]

vertex = ["A", "B", "C", "D"]

key = [999] * 4
parent = [-1] * 4
visited = [False] * 4

key[0] = 0

for i in range(4):

    minimum = 999

    for j in range(4):
        if not visited[j] and key[j] < minimum:
            minimum = key[j]
            u = j

    visited[u] = True

    for v in range(4):
        if graph[u][v] != 0 and not visited[v]:
            if graph[u][v] < key[v]:
                key[v] = graph[u][v]
                parent[v] = u

print("Minimum Spanning Tree")
print("Edge\tWeight")

total = 0
for i in range(1, 4):
    print(vertex[parent[i]], "-", vertex[i], "\t", graph[parent[i]][i])
    total += graph[parent[i]][i]

print("Total Weight =", total)

print("\nMST from A")
print("A")
print("A -> B")
print("A -> B -> C")
print("A -> B -> D")
```
